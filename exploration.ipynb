# Feature Engineering

```python
X_engineered = X.copy()

X_engineered["hr_sin"] = np.sin(
    2 * np.pi * X_engineered["hr"] / 24
)

X_engineered["hr_cos"] = np.cos(
    2 * np.pi * X_engineered["hr"] / 24
)
```

```python
X_engineered[
    ["hr", "hr_sin", "hr_cos"]
].head(24)
```

### Confirm chronological order

```python
df[["dteday", "hr"]].head()
```

```python
df[["dteday", "hr"]].tail()
```

```python
df["dteday"] = pd.to_datetime(df["dteday"])

df = df.sort_values(
    ["dteday", "hr"]
).reset_index(drop=True)
```

### recreate the engineered features

```python
features_to_drop = [
    "instant",
    "dteday",
    "casual",
    "registered",
    "cnt"
]

X = df.drop(columns=features_to_drop)
y = df["cnt"]

X_engineered = X.copy()

X_engineered["hr_sin"] = np.sin(
    2 * np.pi * X_engineered["hr"] / 24
)

X_engineered["hr_cos"] = np.cos(
    2 * np.pi * X_engineered["hr"] / 24
)
```

### Calculate the split point

```python
split_index = int(len(df) * 0.8)

print("Total observations:", len(df))
print("Split index:", split_index)
```

```python
X_train = X_engineered.iloc[:split_index]
X_test = X_engineered.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Training period:")
print(df.iloc[:split_index]["dteday"].min())
print("to")
print(df.iloc[:split_index]["dteday"].max())

print()

print("Testing period:")
print(df.iloc[split_index:]["dteday"].min())
print("to")
print(df.iloc[split_index:]["dteday"].max())
```

```python
from sklearn.linear_model import LinearRegression
```

```python
linear_model = LinearRegression()
```

```python
linear_model.fit(X_train, y_train)
```

```python
y_pred_linear = linear_model.predict(X_test)
```

```python
comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred_linear
})

comparison.head(10)
```

### Evaluate the model

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
mae_linear = mean_absolute_error(
    y_test,
    y_pred_linear
)

rmse_linear = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_linear
    )
)

r2_linear = r2_score(
    y_test,
    y_pred_linear
)

print("Linear Regression")
print("-------------------")
print(f"MAE:  {mae_linear:.2f}")
print(f"RMSE: {rmse_linear:.2f}")
print(f"R²:   {r2_linear:.3f}")
```

### Actual vs Predicted

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred_linear,
    alpha=0.4
)

plt.xlabel("Actual Bike Demand")
plt.ylabel("Predicted Bike Demand")
plt.title("Linear Regression: Actual vs Predicted")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred_linear,
    alpha=0.4
)

min_value = min(y_test.min(), y_pred_linear.min())
max_value = max(y_test.max(), y_pred_linear.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    "--"
)

plt.xlabel("Actual Bike Demand")
plt.ylabel("Predicted Bike Demand")
plt.title("Linear Regression: Actual vs Predicted")

plt.show()
```

```python
residuals = y_test.values - y_pred_linear
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    y_pred_linear,
    residuals,
    alpha=0.4
)

plt.axhline(0, linestyle="--")

plt.xlabel("Predicted Bike Demand")
plt.ylabel("Residual")
plt.title("Linear Regression Residual Plot")

plt.show()
```
